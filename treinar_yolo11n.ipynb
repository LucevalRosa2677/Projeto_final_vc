```python
import zipfile
from pathlib import Path

zip_path = Path("/content/drive/MyDrive/Sentinela/sentinela_para_colab.zip")
with zipfile.ZipFile(zip_path) as arquivo:
    arquivo.extractall("/content")
print("imagens de treino:", len(list(Path("/content/datasets/sentinela/images/train").glob("*"))))
```

```python
from ultralytics import YOLO

modelo = YOLO("yolo11n.pt")
modelo.train(
    data="/content/data.yaml",
    epochs=40,
    imgsz=640,
    batch=16,
    patience=15,
    project="/content/runs",
    name="sentinela",
    exist_ok=True,
)
```

```python
metricas = modelo.val()
print("mAP50:", round(float(metricas.box.map50), 4))
print("mAP50-95:", round(float(metricas.box.map), 4))
```
